```python
# 3단계: 학습 (레이즈 제한 없음). 30분마다 진행 상황을 출력해요
import subprocess, time
os.chdir("/kaggle/working")
env = dict(os.environ, MP_MAX_RAISES="0")
log = open("/kaggle/working/train_mp6rfree.log", "w")
proc = subprocess.Popen(["python", "-u", "mp_cfr.py", "6", str(MINUTES), "--no-report"],
                        env=env, stdout=log, stderr=subprocess.STDOUT)
print("학습 시작")

def last_line(path):
    try:
        lines = [l for l in open(path, encoding="utf-8", errors="ignore").read().splitlines() if l.strip()]
        return lines[-1] if lines else "(아직 출력 없음)"
    except FileNotFoundError:
        return "(로그 없음)"

while proc.poll() is None:
    time.sleep(1800)
    print(last_line("/kaggle/working/train_mp6rfree.log"))
print("끝! 종료 코드:", proc.returncode)
```
